```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside the notebook
%matplotlib inline

print("Libraries imported successfully!")
```

```python
# Load the dataset

df = pd.read_csv("student_dropout_behavior_dataset-selected-columns.csv")

print("Dataset loaded successfully!")
```

```python
# Display the first five rows of the dataset

df.head()
```

```python
# Display the number of rows and columns

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display information about the dataset

df.info()
```

```python
# Count missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Count duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Display the data type of each column

print(df.dtypes)
```

```python
# Create a copy of the original dataset before cleaning

df_clean = df.copy()

# Remove the column that contains 100% missing values

df_clean.drop(columns=["assignments_submitted"], inplace=True)

# Remove any duplicate rows
df_clean.drop_duplicates(inplace=True)

print("Dataset cleaned successfully!")
```

```python
# Check the shape of the cleaned dataset

print("Rows after cleaning:", df_clean.shape[0])
print("Columns after cleaning:", df_clean.shape[1])
```

```python
# Check for remaining missing values

print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Check for duplicate rows after cleaning

print("Duplicate rows after cleaning:", df_clean.duplicated().sum())
```

```python
# Generate descriptive statistics for numerical columns

df_clean.describe()
```

```python
# Display all column names in the cleaned dataset

print(df_clean.columns.tolist())
```

```python
# Fill missing values in assignments_submitted with the median

median_value = df["assignments_submitted"].median()

df_clean = df.copy()

df_clean["assignments_submitted"] = df_clean["assignments_submitted"].fillna(median_value)

print("Missing values filled successfully.")
print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Fill missing values using the median

df_clean["assignments_submitted"] = df_clean["assignments_submitted"].fillna(
    df_clean["assignments_submitted"].median()
)

print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Remove duplicate rows from the cleaned dataset

df_clean = df_clean.drop_duplicates()

print("Duplicate rows removed.")
print("Number of duplicate rows remaining:", df_clean.duplicated().sum())
```

```python
# Check the shape of the cleaned dataset

print("Cleaned dataset shape:", df_clean.shape)
```

```python
# Generate descriptive statistics for numerical columns

print("Descriptive Statistics:")
display(df_clean.describe())
```

```python
# Calculate average marks for each quiz

quiz_means = df_clean[["quiz1_marks", "quiz2_marks", "quiz3_marks"]].mean()

print("Average marks for each quiz:")
print(quiz_means)
```

```python
# Compare average performance across quizzes

quiz_means = df_clean[["quiz1_marks", "quiz2_marks", "quiz3_marks"]].mean()

plt.figure(figsize=(8, 5))

quiz_means.plot(kind="bar")

plt.title("Average Marks Across Quizzes")
plt.xlabel("Quiz")
plt.ylabel("Average Marks")

plt.xticks(rotation=0)
plt.show()
```

```python
# Analyze assignment submission behavior

print("Average total assignments:",
      round(df_clean["total_assignments"].mean(), 2))

print("Average assignments submitted:",
      round(df_clean["assignments_submitted"].mean(), 2))

print("Average submission rate (%):",
      round(
          (df_clean["assignments_submitted"] /
           df_clean["total_assignments"] * 100).mean(), 2
      ))
```

```python
# Visualize relationship between Quiz 1 marks and Midterm marks

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="quiz1_marks",
    y="midterm_marks"
)

plt.title("Quiz 1 Marks vs Midterm Marks")
plt.xlabel("Quiz 1 Marks")
plt.ylabel("Midterm Marks")

plt.show()
```

```python
# Fill missing assignment submissions with 0
df_clean["assignments_submitted"] = df_clean["assignments_submitted"].fillna(0)

# Calculate assignment submission rate
df_clean["submission_rate"] = (
    df_clean["assignments_submitted"] /
    df_clean["total_assignments"]
) * 100

# Display the submission rate
print(df_clean[["student_id", "assignments_submitted", "total_assignments", "submission_rate"]].head())
```

```python
# Visualize the distribution of Quiz 1 marks

plt.figure(figsize=(8, 5))

sns.histplot(
    df_clean["quiz1_marks"],
    bins=10,
    kde=True
)

plt.title("Distribution of Quiz 1 Marks")
plt.xlabel("Quiz 1 Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Visualize relationship between quiz marks and midterm marks

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="quiz1_marks",
    y="midterm_marks"
)

plt.title("Quiz 1 Marks vs Midterm Marks")
plt.xlabel("Quiz 1 Marks")
plt.ylabel("Midterm Marks")

plt.show()
```

```python
# Count students in each gender category

gender_counts = df_clean["gender"].value_counts()

print(gender_counts)
```

```python
# Create a bar chart of gender distribution

plt.figure(figsize=(8, 5))

gender_counts.plot(kind="bar")

plt.title("Gender Distribution of Students")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)

plt.show()
```

```python
# Display the final cleaned dataset information

print("Final dataset shape:", df_clean.shape)
print("\nFinal data types:")
print(df_clean.dtypes)

print("\nRemaining missing values:")
print(df_clean.isnull().sum())

print("\nRemaining duplicate rows:", df_clean.duplicated().sum())
```

```python
# Check the values in assignments_submitted

print("Assignments submitted:")
print(df_clean["assignments_submitted"].value_counts(dropna=False))

print("\nSummary:")
print(df_clean["assignments_submitted"].describe())
```
